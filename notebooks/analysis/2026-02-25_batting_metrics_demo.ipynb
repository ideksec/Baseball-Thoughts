# How do core batting metrics compare across a sample of players?

| Field | Value |
|-------|-------|
| **Question** | How do AVG, SLG, OBP, and OPS distribute across a small sample of hitters? |
| **Datasets** | Inline sample data (fictional, for framework demo) |
| **Time window** | N/A — synthetic data |
| **Assumptions** | Counting stats are pre-aggregated per player; SF and HBP included |

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from baseball_lab.metrics.batting import batting_avg, slugging_pct, on_base_pct, ops

sns.set_theme(style="whitegrid")
%matplotlib inline
```

## Method

We build a small DataFrame of fictional player seasons, compute four standard batting
metrics using `baseball_lab.metrics.batting`, and visualize the results.

```python
# Sample data — fictional players
data = [
    {"player": "Alice",  "AB": 500, "1B": 90, "2B": 25, "3B": 4, "HR": 18, "BB": 55, "HBP": 5, "SF": 4},
    {"player": "Bob",    "AB": 450, "1B": 70, "2B": 30, "3B": 2, "HR": 30, "BB": 60, "HBP": 8, "SF": 3},
    {"player": "Carol",  "AB": 520, "1B": 110, "2B": 20, "3B": 6, "HR": 5, "BB": 40, "HBP": 2, "SF": 5},
    {"player": "Dan",    "AB": 480, "1B": 80, "2B": 28, "3B": 3, "HR": 22, "BB": 50, "HBP": 3, "SF": 6},
    {"player": "Eve",    "AB": 400, "1B": 60, "2B": 15, "3B": 1, "HR": 35, "BB": 70, "HBP": 10, "SF": 2},
]

df = pd.DataFrame(data)
df["H"] = df["1B"] + df["2B"] + df["3B"] + df["HR"]

# Compute metrics using our reusable functions
df["AVG"] = df.apply(lambda r: batting_avg(r["H"], r["AB"]), axis=1)
df["SLG"] = df.apply(lambda r: slugging_pct(r["1B"], r["2B"], r["3B"], r["HR"], r["AB"]), axis=1)
df["OBP"] = df.apply(lambda r: on_base_pct(r["H"], r["BB"], r["HBP"], r["AB"], r["SF"]), axis=1)
df["OPS"] = df.apply(lambda r: ops(r["1B"], r["2B"], r["3B"], r["HR"], r["BB"], r["HBP"], r["AB"], r["SF"]), axis=1)

df[["player", "AB", "H", "AVG", "SLG", "OBP", "OPS"]].round(3)
```

## Results

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Bar chart: OPS by player
sns.barplot(data=df, x="player", y="OPS", ax=axes[0], palette="viridis")
axes[0].set_title("OPS by Player")
axes[0].set_ylabel("OPS")

# Scatter: OBP vs SLG
sns.scatterplot(data=df, x="OBP", y="SLG", hue="player", s=120, ax=axes[1])
axes[1].set_title("OBP vs SLG")

plt.tight_layout()
plt.show()
```

## Conclusion

This notebook demonstrates the framework by computing core batting metrics (AVG, SLG, OBP, OPS)
using reusable functions from `baseball_lab.metrics.batting`. The sample data is synthetic —
no external data sources were needed.

Eve shows the highest OPS despite a modest batting average, driven by power (35 HR) and
strong plate discipline (70 BB, 10 HBP). Bob follows a similar pattern. Carol has the highest
AVG but the lowest OPS, reflecting a contact-heavy, low-power profile.

- **Finding:** OPS captures the power-vs-contact tradeoff better than AVG alone.
- **Limitations:** Synthetic data; real-world distributions would show more variance.
